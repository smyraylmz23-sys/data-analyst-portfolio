```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# Gereksiz uyarı mesajlarını gizliyor ve grafiklerin genel temasını ayarlıyoruz
warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (11, 6)
plt.rcParams["axes.titlesize"] = 13
plt.rcParams["axes.labelsize"] = 11

print(" Kütüphaneler başarıyla yüklendi ve grafik ortamı hazırlandı.")
```

## 1. Veri Kaynaklarının Yüklenmesi, Temizlenmesi ve Profilleme

İki bağımsız kurumsal kaynaktan beslenen veri setlerimizin yapısal anatomisi (satır-sütun boyutları, veri tipleri, eksik kayıtlar ve örnek gözlemler) incelenmektedir.

```python
# 1. İTO Yeni Şirket Kuruluş Verisinin Okunması ve Temizlenmesi
ito_dosya = "/content/sample_data/tbl_itokurulus.xlsx"
tablo_ito = pd.read_excel(ito_dosya)

# NACE Kodu veya Şirket Sayısı boş olan satırları temizliyoruz
tablo_ito = tablo_ito.dropna(subset=["Nace Kodu", "Şirket Sayısı"]).copy()
tablo_ito["Şirket Sayısı"] = tablo_ito["Şirket Sayısı"].astype(int)

# 2. TBB Makro Kredi Raporunu Okuyoruz
tbb_dosya = "/content/sample_data/Seçilmiş Göstergelerin Coğrafi Dağılımı 2025.xls"
tablo_kredi_makro = pd.read_excel(tbb_dosya, sheet_name="t13")
ist_ticari_kredi = float(tablo_kredi_makro.iloc[5, 7])

print("=" * 65)
print(f"İTO Veri Seti Boyutu : {tablo_ito.shape[0]} Sektör (Satır), {tablo_ito.shape[1]} Sütun")
print(f"Toplam Tescil Edilen Şirket: {tablo_ito['Şirket Sayısı'].sum():,} Adet")
print(f"İstanbul Toplam Ticari Kredi Hacmi: {ist_ticari_kredi:,.0f} Milyon TL")
print("=" * 65)

# Eksik Değer Kontrolü
eksik_sayisi = tablo_ito.isnull().sum().sum()
print(f"İTO Veri Setinde Eksik (NaN) Kayıt Sayısı: {eksik_sayisi}")

# İlk 5 Gözlem (head)
print("\n İTO Veri Setinden İlk 5 Satır:")
display(tablo_ito.head(5))

# Betimsel İstatistikler
print("\n Şirket Sayıları Betimsel İstatistik Özeti:")
display(tablo_ito[["Şirket Sayısı"]].describe().round(1).T)
```

## 2. Girişim Dağılımı ve Aykırı Değer Analizi (Skewness & Outlier Detection)

İstanbul'da tescil edilen şirketlerin sektörlere homojen dağılıp dağılmadığını, ortalama-medyan ilişkisi ve **Kutu Grafiği (Boxplot)** yardımıyla inceliyoruz.

```python
medyan_sirket = tablo_ito["Şirket Sayısı"].median()
ortalama_sirket = tablo_ito["Şirket Sayısı"].mean()

fig, ax = plt.subplots(figsize=(10, 4))
sns.boxplot(x=tablo_ito["Şirket Sayısı"], color="#3498db", fliersize=5, ax=ax)

ax.set_title("İTO Sektörel Şirket Kuruluş Dağılımı ve Aykırı Değerler (Boxplot)", fontsize=13, fontweight="bold")
ax.set_xlabel("Sektör Başına Düşen Şirket Sayısı", fontsize=11)
plt.tight_layout()
plt.show()

print(f" Ortalama Şirket Sayısı : {ortalama_sirket:.1f}")
print(f" Medyan Şirket Sayısı   : {medyan_sirket:.1f}")
print(" Çıkarım: Ortalama değer medyandan belirgin şekilde büyüktür. Bu durum, veri setinin aşırı sağa çarpık olduğunu ve girişimlerin tabana yayılmayıp belirli birkaç ana sektörde (İnşaat, Yazılım, E-Ticaret) aşırı yığıldığını kanıtlar.")
```

## 3. Kredi Arz Tarafı Analizi: Banka Mülkiyet Kırılımı ve Pazar Payları

İstanbul'daki toplam ticari kredi likiditesinin Kamu, Yerli Özel ve Yabancı Sermayeli bankalar arasındaki dağılımını hesaplıyor; likidite yükünü pazarın hangi aktörünün çektiğini **Donut (Halka) Grafik** ile görselleştiriyoruz.

> **Metodolojik Not:** Çifte sayım (double counting) riskini önlemek amacıyla konsolide `Tüm Bankalar.xlsx` ve `Mevduat Bankaları.xlsx` dosyaları yerine doğrudan mülkiyet alt kırılımları (`Kamusal`, `Özel`, `Yabancı`) kullanılmıştır.

```python
import matplotlib.pyplot as plt

# TBB 2025 Yıllık Raporu - Tablo 25 (İstanbul İli Net Ticari Kredi Hacimleri - Milyon TL)
kamu_kredi = 2511758.0      # Kamusal Sermayeli Mevduat Bankaları
ozel_kredi = 2838214.0      # Özel Sermayeli Mevduat Bankaları
yabanci_kredi = 1970428.0   # Yabancı Sermayeli Mevduat Bankaları

toplam_kredi = kamu_kredi + ozel_kredi + yabanci_kredi
kamu_oran = (kamu_kredi / toplam_kredi) * 100
ozel_oran = (ozel_kredi / toplam_kredi) * 100
yabanci_oran = (yabanci_kredi / toplam_kredi) * 100

print("-" * 55)
print(f"Kamu Bankaları   : {kamu_kredi:,.0f} Milyon TL (%{kamu_oran:.1f})")
print(f"Özel Bankalar    : {ozel_kredi:,.0f} Milyon TL (%{ozel_oran:.1f})")
print(f"Yabancı Bankalar : {yabanci_kredi:,.0f} Milyon TL (%{yabanci_oran:.1f})")
print("-" * 55)

# Donut Grafik Çizimi
etiketler = ["Kamu Bankaları", "Yerli Özel Bankalar", "Yabancı Sermayeli Bankalar"]
oranlar = [kamu_kredi, ozel_kredi, yabanci_kredi]
renkler = ["#d62728", "#1f77b4", "#ff7f0e"]

fig, ax = plt.subplots(figsize=(8, 5.5))
ax.pie(
    oranlar,
    labels=etiketler,
    colors=renkler,
    autopct="%1.1f%%",
    startangle=90,
    explode=(0.04, 0.04, 0.04),
    textprops=dict(color="black", fontsize=10, fontweight="bold")
)

# Ortadaki beyaz daire (Donut efekti)
merkez_daire = plt.Circle((0, 0), 0.70, fc="white")
fig.gca().add_artist(merkez_daire)

plt.title("İstanbul'da Banka Mülkiyetine Göre Kredi Dağılımı ve Pazar Payı (2025)", fontsize=13, fontweight="bold", pad=15)
plt.tight_layout()
plt.show()
```

## 4. Talep Tarafı Analizi: En Yüksek Tescil Hacmine Sahip İlk 10 Sektör

Girişim talebinin sektörel odaklarını ortaya çıkarmak için İTO veri setindeki tescil sayılarına göre lider ilk 10 NACE sektörü yatay çubuk grafik ile görselleştirilmektedir.

```python
ilk10_sektor = tablo_ito.sort_values(by="Şirket Sayısı", ascending=False).head(10)

tanim_sutun_adaylari = [col for col in ilk10_sektor.columns if any(k in col.lower() for k in ["tanım", "başlık", "baslik"])]
tanim_sutunu = tanim_sutun_adaylari[0] if len(tanim_sutun_adaylari) > 0 else "Nace Kodu"

kisaltilmis_etiketler = [
    f"{str(satir[tanim_sutunu])[:35]}... ({satir['Nace Kodu']})"
    for _, satir in ilk10_sektor.iterrows()
]

fig, ax = plt.subplots(figsize=(11, 6.5))
cubuklar = ax.barh(kisaltilmis_etiketler[::-1], ilk10_sektor["Şirket Sayısı"][::-1], color="#1f77b4", edgecolor="black")

for cubuk in cubuklar:
    w = cubuk.get_width()
    ax.text(w + 50, cubuk.get_y() + cubuk.get_height()/2, f"{int(w):,} adet", va="center", fontsize=9.5, fontweight="bold")

plt.title("İstanbul'da En Çok Kurulan İlk 10 Sektör (Talep Tarafı - 2025/2026)", fontsize=13, fontweight="bold", pad=15)
plt.xlabel("Yeni Kurulan Şirket Sayısı", fontsize=11)
plt.ylabel("NACE Sektör Tanımı ve Kodu", fontsize=11)
plt.xlim(0, ilk10_sektor["Şirket Sayısı"].max() * 1.15)
plt.tight_layout()
plt.show()
```

## 5. Finansman Boşluğu (Funding Gap) Analizi: Geleneksel vs. Yeni Nesil Girişimler

Geleneksel fiziki teminatlı sektörler (İnşaat, Yeme-İçme) ile teminatsız büyüyen yeni nesil dijital sektörler (Yazılım, E-Ticaret, Yönetim Danışmanlığı) hacimsel olarak karşılaştırılmaktadır.

```python
tablo_ito["Nace_Str"] = tablo_ito["Nace Kodu"].astype(str).str.strip()

geleneksel_kodlar = ["41", "41.0", "56.11"]
dijital_kodlar = ["62.1", "47.91", "70.2"]

geleneksel_toplam = tablo_ito[tablo_ito["Nace_Str"].isin(geleneksel_kodlar)]["Şirket Sayısı"].sum()
dijital_toplam = tablo_ito[tablo_ito["Nace_Str"].isin(dijital_kodlar)]["Şirket Sayısı"].sum()

kategoriler = [
    "Geleneksel Sektörler\n(İnşaat, Lokanta)\nFiziki Teminat Odaklı",
    "Yeni Nesil Dijital Sektörler\n(Yazılım, E-Ticaret, Danışmanlık)\nNakit Akışı Odaklı"
]
degerler = [geleneksel_toplam, dijital_toplam]
renkler = ["#7f7f7f", "#2ca02c"]

fig, ax = plt.subplots(figsize=(9, 5.5))
cubuklar = ax.bar(kategoriler, degerler, color=renkler, width=0.45, edgecolor="black")

toplam_iki_grup = geleneksel_toplam + dijital_toplam
for cubuk in cubuklar:
    h = cubuk.get_height()
    oran = (h / toplam_iki_grup) * 100
    ax.text(cubuk.get_x() + cubuk.get_width()/2, h + 120, f"{int(h):,} Şirket\n(%{oran:.1f})", ha="center", va="bottom", fontsize=11, fontweight="bold")

plt.title("Geleneksel vs. Yeni Nesil Şirket Kuruluş Hacimleri (Kredi Uyuşmazlığı Potansiyeli)", fontsize=13, fontweight="bold", pad=20)
plt.ylabel("Yeni Kurulan Şirket Sayısı", fontsize=11)
plt.ylim(0, max(degerler) * 1.25)
plt.tight_layout()
plt.show()
```

### 6.2. Derinlemesine Analitik Değerlendirme

#### 1. Bankacılık Sektörünün Kredi Arz Mimarisi ve Risk Algısı
İstanbul'daki bankacılık mevduat bankaları kredi pastasında **Yerli Özel Bankaların %38,8** ve **Kamu Bankalarının %34,3** oranındaki payları, sistemin iki ana sütun üzerine oturduğunu göstermektedir. Ancak özel ve yabancı sermayeli bankalar; Basel kriterleri, sermaye yeterlilik rasyoları (SYR) ve aktif kalitesi baskıları nedeniyle kredi tahsis süreçlerinde son derece ihtiyatlı davranmaktadır.

Bankalar kredi verirken birinci geri ödeme kaynağı olarak "nakit akışını" kabul etseler de, temerrüt (NPL) riskine karşı ikincil güvence olarak mutlaka **likit ve fiziki teminat (gayrimenkul ipoteği, fabrika, arsa)** talep etmektedir. Bu durum, kredinin reel sektöre dağılımında yapısal bir bariyer oluşturmaktadır.

#### 2. Reel Sektör Girişim Talebinin Dinamikleri ve Sektörel Kutuplaşma
İTO verilerinde en çok kurulan ilk 10 sektör incelendiğinde ikili bir yapı dikkat çekmektedir:
* **Geleneksel Çekirdek:** Bina İnşaatı (**4.574 adet**) ve Lokantacılık (**2.459 adet**) tescil adetlerinde liderdir. İnşaat sektörü, doğası gereği yüksek fiziki sermaye gerektiren ve arkasında somut bir gayrimenkul varlığı bırakan bir alandır.
* **Hızla Büyüyen Yeni Ekonomi:** Bilgisayar Programlama / Yazılım (**2.229 adet**), Perakende E-Ticaret (**1.509 adet**) ve İşletme Danışmanlığı (**874 adet**) gibi alanlar toplamda **4.612 yeni şirkete** ulaşmıştır. Bu rakam, odaklanan şirket grubunun yaklaşık **%40'ını** oluşturmaktadır.

#### 3. Finansman Boşluğu Teorisinin Doğrulanması (Funding Gap Hypothesis)
Elde edilen sonuçlar, projenin çıkış noktasındaki hipotezi doğrulamaktadır:
* İstanbul'da her 10 yeni girişimden yaklaşık 4'ü dijital, entelektüel sermayeye ve e-ticaret altyapısına dayalı olarak kurulmaktadır.
* Bu şirketlerin bilançosunda makine, arsa veya bina gibi bankacılık regülasyonlarının "teminat" kabul ettiği maddi duran varlıklar bulunmamaktadır. Şirketin ana değeri patent, yazılım kodu, müşteri portföyü ve yinelenen abonelik/ciro geliridir.
* Bankacılık sisteminin **8,1 Trilyon TL'lik** devasa kredi arzı bulunmasına rağmen, bu fonlar geleneksel teminat şartları nedeniyle dijital girişimcilere akamamakta; girişimciler öz sermaye yetersizliği ve erken aşama nakit sıkışıklığı ("Ölüm Vadisi / Valley of Death") ile karşı karşıya kalmaktadır.

### 6.3. Sektörel ve Politika Düzeyinde Stratejik Öneriler

Bu finansman tıkanıklığının aşılması ve bankacılık likiditesinin yüksek katma değerli dijital alanlara kanalize edilebilmesi için şu stratejiler geliştirilmelidir:

1. **Nakit Akışına Dayalı Kredi Skorlama Modelleri (Cash-Flow Lending):**
   * Bankalar, geleneksel bilanço ve mülkiyet ipoteği yerine; girişimlerin POS ciroları, e-ticaret platform satış hacimleri, e-fatura hareketleri ve düzenli abonelik gelirlerini (ARR/MRR) analiz eden yapay zeka destekli dinamik skorlama algoritmalarına geçmelidir.

2. **Kredi Garanti Fonu (KGF) ve Hazine Destekli Kefalet Revizyonu:**
   * Teminatsız dijital şirketlerin banka kredisine erişebilmesi için KGF mekanizması bünyesinde "Yazılım ve Teknoloji Girişimleri Özel Kefalet Paketi" tanımlanmalı, bankanın maruz kaldığı kredi riski kamusal kefaletle paylaşılmalıdır.

3. **Girişim Sermayesi ve Girişim Borcu (Venture Debt) Entegrasyonu:**
   * Bankalar, yüksek büyüme vadeden ancak teminatı olmayan teknoloji şirketlerine doğrudan standart ticari kredi vermek yerine; hisse opsiyonlu (warrant) girişim borcu ürünleri geliştirmeli veya kendi kurumsal girişim sermayesi (CVC) fonları üzerinden borç-özkaynak hibrit finansman modelleri sunmalıdır.

4. **Fikri ve Sınai Mülkiyetin (IP) Teminatlaştırılması:**
   * Türkiye'deki yasal mevzuat altyapısı geliştirilerek yazılım telifleri, tescilli algoritmalar ve patentlerin bankalar nezdinde taşınır rehni veya teminat unsuru olarak tescil edilebilmesi sağlanmalıdır.

# **Araştırma Soruları Cevapları:**

**Soru 1:** İstanbul’daki yeni şirketleşme talebinin genel büyüklüğü nedir ve veri analize hazır mı?

**Cevap:** İTO veri setinde toplam 42.382 şirketin tescil edildiği, 566 farklı NACE sektörüne yayıldığı ve analizimizi bozacak eksik (NaN) değerlerin başarıyla temizlendiği gösterildi.

**Soru 2:** Girişimler sektörlere dengeli mi dağılıyor, yoksa belirli alanlarda aşırı yığılma (kümelenme) var mı?

**Cevap:** Kutu Grafiği (Boxplot) ve ortalama-medyan kıyaslamasıyla cevaplandı. Sektör başına düşen ortalama şirket sayısı medyandan belirgin şekilde yüksek çıktı. Bu durum, veri setinin aşırı sağa çarpık (skewed) olduğunu ve girişimlerin tabana yayılmak yerine birkaç baskın sektörde aşırı yığıldığını matematiksel olarak ispatladı.

**Soru 3:** İstanbul’daki devasa ticari kredi likiditesini (yaklaşık 7,3 - 8,1 Trilyon TL) hangi banka türleri finanse ediyor?

**Cevap:** Donut (Halka) Grafik ile pazar payları netleştirildi:

Yerli Özel Bankalar: %38,8 pay (2,8 Trilyon TL) ile pazar lideri ve likiditenin ana taşıyıcısı.

Kamu Bankaları: %34,3 pay (2,5 Trilyon TL) ile dengeli ikinci güç.

Yabancı Sermayeli Bankalar: %26,9 pay (1,9 Trilyon TL) ile kritik bir uluslararası fon sağlayıcısı.

**Soru 4:** Bankacılık sisteminde likidite riski kamuya mı yoksa özel sektöre mi yığılmış durumda?

**Cevap:** Özel ve yabancı bankaların toplam payının %65'in üzerinde olduğu, dolayısıyla piyasa likiditesinin ve risk algısının ağırlıklı olarak özel piyasa dinamiklerine ve sıkı Basel/kredi tahsis kurallarına bağlı olduğu görüldü.


**Soru 5:** İstanbul'da en çok kurulan ve girişimcilerin en yoğun talep gösterdiği sektörler hangileridir?

**Cevap:** Yatay Çubuk Grafik ile ilk 10 NACE kodu sıralandı. İlk 10 sektörün tek başına 15.823 şirketle toplam tescillerin %37,3'ünü oluşturduğu ortaya kondu:

Lider geleneksel sektör: Bina İnşaatı (4.574 adet) ve Lokantacılık (2.459 adet).

Lider yeni nesil sektör: Bilgisayar Programlama / Yazılım (2.229 adet) ve Perakende E-Ticaret (1.509 adet).

**Soru 6:** Bankaların kredi verirken aradığı "fiziki teminat (gayrimenkul/ipotek)" modeli, yeni kurulan şirket yapısıyla uyuşuyor mu?

**Cevap:** Karşılaştırmalı Sütun Grafiği ile cevaplandı. İncelenen odak girişimlerin:

%60,4'ü (7.033 şirket): İnşaat ve yeme-içme gibi bankaların geleneksel ipotek teminatına uygun sektörlerde.

%39,6'sı (4.612 şirket): Yazılım, e-ticaret ve danışmanlık gibi tamamen teminatsız, entelektüel sermaye ve nakit akışına dayalı yeni nesil sektörlerde kuruldu.

**Soru 7:** Bankalarda trilyonlarca liralık kredi hacmi varken neden yeni nesil girişimciler finansmana ulaşamıyor (Finansman Boşluğu Nerede)?

**Cevap:** Her 10 yeni şirketten 4'ünün fiziki ipotek gösterememesi sebebiyle bankacılık regülasyonlarına takıldığı; bankaların bilançodaki maddi duran varlığa bakması yüzünden bu dijital potansiyelin kredi pastasından pay alamayıp "Finansman Boşluğu" (Funding Gap) içinde kaldığı ispatlandı.

**Soru 8:** Bu uyuşmazlığı çözmek için bankacılık sektörü ve ekonomi yönetimi ne yapmalıdır?

**Cevap:** Raporun sonuç kısmında 4 somut stratejik öneriyle yanıtlandı:

İpotek yerine e-fatura ve POS akışına dayalı Nakit Akışı Odaklı Kredi Skorlaması (Cash-Flow Lending).

KGF bünyesinde teminatsız teknoloji girişimlerine özel Kefalet Paketleri.

Bankalar bünyesinde hisse opsiyonlu Girişim Borcu (Venture Debt) ürünleri.

Yazılım telifleri ve patentlerin yasal teminat sayılmasını sağlayacak mevzuat reformu.
